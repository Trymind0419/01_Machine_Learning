## 회귀 분석용 이상치 처리 
- Z-Score 방법으로 이상치 찾기

```python
import numpy as np
```

```python
# 선형(주로 선형 데이터에서 사용)회귀는 정규 분포 가정을 따름으로,
#  - 평균으로부터 표준편차의 몇 배만큼 떨어져 있는지를 나타내는 Z-Score를 사용함
# Z = (X - μ) / σ (X: 데이터 포인트, μ: 평균, σ: 표준편차)
# Z보다 큰(초과) 값들은 이상치로 판단(작은 값이 정상 데이터)


### TotalCharges 데이터에서 이상치 유무 확인하기
# - X = TotalCharges 데이터가 됨
z_score = np.abs((df["TotalCharges"] - df["TotalCharges"].mean()) / df["TotalCharges"].std())
print(z_score)

### 이상치 기준을 결정하기
# Z-Score의 값이 3(약 99.7% 신뢰구간을 의미함) 이상인 데이터를 이상치로 규정함
# Z-Score는 신뢰구간을 나타내는 값을 의미함

### 이상치 제거 전 df 정보 : 전체 5000건
print("----------[Z-Score 이상치 처리 전]-----------")
df.info()

### 이상치 제거 처리 : z_score 값이 3보다 작은 경우의 행인덱스 위치의 값만 추출
###                 : 이상치 제거 효과와 동일

# 데이터 행단위 필터링(조건정리)
df = df[z_score < 3]
print("----------[Z-Score 이상치 처리 후]-----------")
df.info()
```

## 분류 분석용 이상치 처리 - IQR(InterQuartile Range)
- 박스플롯(boxplolt) 기반 이상치 탐지 방식임
- IQR = Q3-Q1
- Max-Out = 75%사분위수값 Q3 + 1.5 * IQR
- Min-Out = 25%사분위수값 Q1- 1.5 * IQR

```python

# 비선형 및 범주형일 경우 IQR사용
# 난 아무것도 모르겠다! IQR로 사용
### Q1 및 Q3 값 추출하기
Q1 = df["Age"].quantile(0.25)
Q3 = df["Age"].quantile(0.75)
print(f"Q1={Q1}, Q3={Q3}")

### IQR 계산하기
IQR = Q3 - Q1
print(f"IQR={IQR}")

### Max-Out = 75%사분위수값 Q3 + 1.5 * IQR
### Min-Out = 25%사분위수값 Q1- 1.5 * IQR
lower_bound = Q1-1.5 * IQR 
upper_bound = Q3+1.5 * IQR
print(f"lower_bound={lower_bound}, upper_bound={upper_bound}")
```

```python
### IQR 이상치 처리 : lower_bound 이상이고, upper_bound 이하인 데이터만 추출하여 df에 넣기
df = df[(df["Age"] >= lower_bound) & (df["Age"] <= upper_bound)] # and 연산자는 괄호써줘야함.

df.info()
```

```python
import pandas as pd
import numpy as np


# 읽어오기
file_path = "./dataset/01_02_ecommerce_customers_noisy.csv"
df = pd.read_csv(file_path) #df는 메모리 주소값임. 데이터프레임으로 받은 클래스의 주소값을 받아옴

# 전처리 함수 생성하기 : 사용자 정의 함수(직접 만든 함수를 의미)
# 함수명 : clean_ecommerce_data
# 처리내용 : 지금까지 했던 전처리 모든 것
# 반환값(return) : 전처리 끝난 데이터프레임

### 함수 정의시 고려 사항
# 1. input data : 어떤 데이터를 받아서 사용할지? (없을 수도 있음) / 매개변수(파라미터) : 함수 () 내 작성 
# 2. 처리 프로그램 : 어떤 처리를 할지? (받아오는 데이터가 있다면 이 데이터를 어떻게 사용할지?)
# 3. output data : 최정 결과 데이터를 넘겨 줄지 말지?


def clean_ecommerce_data(df_one):
# ↓ 처리 프로그램///....
    # Age를 중앙값으로 대체
    age_median = df_one["Age"].median()
    df_one["Age"] = df_one["Age"].fillna(age_median)

    # MonthlyCharges를 평균으로 대체
    mc_mean = df_one["MonthlyCharges"].mean()
    df_one["MonthlyCharges"] = df_one["MonthlyCharges"].fillna(mc_mean)

    # z-score 3이상인 것들 없애기
    z_score = np.abs((df_one["TotalCharges"] - df_one["TotalCharges"].mean()) / df_one["TotalCharges"].std())
    df_one = df_one[z_score < 3]

    # IQR 이상치 제거 [Age]
    Q1 = df_one["Age"].quantile(0.25)
    Q3 = df_one["Age"].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    df_one = df_one[(df_one["Age"] >= lower_bound) & (df_one["Age"] <= upper_bound)]

    # 인덱스 재정의
    df_one.reset_index(inplace=True, drop=True)

    # 나이를 정수형태로
    df_one["Age"] = df_one["Age"].astype("int")

    # 성별을 소문자로
    df_one["Gender"] = df_one["Gender"].str.lower()
    # output data : 반환(return) -> 전처리가 완료된 데이터 프레임 반환
    return df_one

save_path = "./dataset/01_03_clean_ecommerce_data_clean.csv"
df_final=clean_ecommerce_data(df)
df_final.to_csv(save_path, index=False)
```

```python
##### 추가 학습

import pandas as pd
import numpy as np


# 읽어오기
 #df는 메모리 주소값임. 데이터프레임으로 받은 클래스의 주소값을 받아옴

# 전처리 함수 생성하기 : 사용자 정의 함수(직접 만든 함수를 의미)
# 함수명 : clean_ecommerce_data
# 처리내용 : 지금까지 했던 전처리 모든 것
# 반환값(return) : 전처리 끝난 데이터프레임

### 함수 정의시 고려 사항
# 1. input data : 어떤 데이터를 받아서 사용할지? (없을 수도 있음) / 매개변수(파라미터) : 함수 () 내 작성 
# 2. 처리 프로그램 : 어떤 처리를 할지? (받아오는 데이터가 있다면 이 데이터를 어떻게 사용할지?)
# 3. output data : 최정 결과 데이터를 넘겨 줄지 말지?


def clean_ecommerce_data():
    file_path = "./dataset/01_02_ecommerce_customers_noisy.csv"
    df = pd.read_csv(file_path)

# ↓ 처리 프로그램///....
    # Age를 중앙값으로 대체
    age_median = df["Age"].median()
    df["Age"] = df["Age"].fillna(age_median)

    # MonthlyCharges를 평균으로 대체
    mc_mean = df["MonthlyCharges"].mean()
    df["MonthlyCharges"] = df["MonthlyCharges"].fillna(mc_mean)

    # z-score 3이상인 것들 없애기
    z_score = np.abs((df["TotalCharges"] - df["TotalCharges"].mean()) / df["TotalCharges"].std())
    df = df[z_score < 3]

    # IQR 이상치 제거 [Age]
    Q1 = df["Age"].quantile(0.25)
    Q3 = df["Age"].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    df = df[(df["Age"] >= lower_bound) & (df["Age"] <= upper_bound)]

    # 인덱스 재정의
    df.reset_index(inplace=True, drop=True)

    # 나이를 정수형태로
    df["Age"] = df["Age"].astype("int")

    # 성별을 소문자로
    df["Gender"] = df["Gender"].str.lower()
    # output data : 반환(return) -> 전처리가 완료된 데이터 프레임 반환
    return df

clean_ecommerce_data()
```
