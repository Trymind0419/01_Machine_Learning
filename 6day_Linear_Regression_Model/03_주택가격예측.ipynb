```python
# 숫자 처리
import numpy as np

# 데이터 처리
import pandas as pd

# 훈련:검증:테스트 데이터 분리 처리
from sklearn.model_selection import train_test_split

# 훈련모델
# 선형회귀 모델
from sklearn.linear_model import LinearRegression
# KNN 모델
from sklearn.neighbors import KNeighborsRegressor
# 특성 늘리는 클래스
from sklearn.preprocessing import PolynomialFeatures

# 표준화 데이터 스케일링
from sklearn.preprocessing import StandardScaler

# 오차확인(테스트 오차)
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
df = pd.read_csv("./dataset/01_회귀_주택가격예측_데이터셋_Row_1000.csv")
df.info()
df.describe()
df.head(3)
```

```python
# 독립변수(X) : 면적, 층수, 역까지_도보시간, 주변_편의시설_점수
# 종속변수(y) : 주택_가격
```

```python
df_c = df.copy()
df_x = df_c.drop(columns=["주택_가격"], axis=1)
df_x.info()
df_y = df_c["주택_가격"]
df_y.info()
train_X, test_X, train_y, test_y = train_test_split(df_x, df_y, test_size=0.3, random_state=42)
print(train_X.shape, train_y.shape)
print(test_X.shape, test_y.shape)
```

```python
# 데이터 분리하기 7:3

train_X, test_X, train_y, test_y = train_test_split(df_x, df_y, test_size=0.3, random_state=42)
print(train_X.shape, train_y.shape)
print(test_X.shape, test_y.shape)
```

```python
knr = KNeighborsRegressor()
knr

# 훈련시키기
knr.fit(train_X, train_y)

# 정확도()
train_score = knr.score(train_X, train_y)
test_score = knr.score(test_X, test_y)
gap = train_score - test_score
print(train_score, test_score, gap)

# 예측
pred_train = knr.predict(train_X)
pred_test = knr.predict(test_X)
print(pred_train)
print(pred_test)
```

```python
# 클래스 생성
lr = LinearRegression()
lr

# 훈련시키기
lr.fit(train_X, train_y)

# 정확도()
train_score = lr.score(train_X, train_y)
test_score = lr.score(test_X, test_y)
gap = train_score - test_score
print(train_score, test_score, gap)

# 예측
pred_train = lr.predict(train_X)
pred_test = lr.predict(test_X)
print(pred_train)
print(pred_test)
```

```python
std_scaler = StandardScaler()
std_scaler

cols = ["면적", "층수", "주변_편의시설_점수", "역까지_도보시간"]

df[[col + "_std" for col in cols]] = std_scaler.fit_transform(df[cols])
df
```

```python

df_c = df.copy()
df_x = df_c.drop(columns=["주택_가격", "면적", "층수", "역까지_도보시간", "주변_편의시설_점수"], axis=1)
df_x.info()
df_y = df_c["주택_가격"]
df_y.info()
# 독립변수 : 면적_std, 층수_std, 주변_편의시설_점수_std
# 종속변수 : 주택가격
```

```python
# 클래스 생성
knr = KNeighborsRegressor()
knr

# 훈련시키기
knr.fit(train_X, train_y)

# 정확도()
train_score = knr.score(train_X, train_y)
test_score = knr.score(test_X, test_y)
gap = train_score - test_score
print(train_score, test_score, gap)

# 예측
pred_train = knr.predict(train_X)
pred_test = knr.predict(test_X)
print(pred_train)
print(pred_test)
```
